```python
import os, requests, json
from smolagents import LiteLLMModel,DuckDuckGoSearchTool,VisitWebpageTool,CodeAgent,ToolCallingAgent,tool
```

# --- 1. CONFIGURATION DU MODÈLE LOCAL (OLLAMA) ---

```python
model = LiteLLMModel(
    model_id="ollama/llama3.2:latest",
    temperature=0.2,
    timeout=3600  # 1h max
)
```

# --- 2. CONFIGURATION DES OUTILS ---

```python
searchtool = DuckDuckGoSearchTool()
visitTool = VisitWebpageTool()
```

```python
@tool
def custom_tool(word:str)-> str:
    """Renvoie un synonyme français d'un mot (word:str)-> str

    Args:
        word (str): le mot

    Returns:
        str: le synonyme du mot
    """
    
    r = requests.get(
        f"https://api.datamuse.com/words?rel_syn={word}&v=fr"
    )
    data = r.json()
    return data[0]['word'] if data else "Aucun synonyme trouvé"
    
```

# --- 3. INITIALISATION DE L'AGENT ---

```python

""" 
# approche trop lourd et fait planté le pc
code_agent = CodeAgent(
    model = model,
    tools = [searchtool, visitTool, custom_tool],
    add_base_tools = True,
    # IMPORTANT : Autoriser requests pour que l'agent puisse l'utiliser dans son code
    additional_authorized_imports=["requests", "json"]
)
"""

tc_agent = ToolCallingAgent(
    model = model,
    tools = [searchtool, visitTool, custom_tool],
    add_base_tools = False,
)
```
